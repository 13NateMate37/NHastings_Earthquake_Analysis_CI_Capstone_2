# Section 2 - Transformations

The "date_time" columns' data type needs to be changed to DateTime.

```python
# Before
print("Column type is: ")
print(df_raw["date_time"].dtypes)

# Conversion
df_raw["date_time"] = pd.to_datetime(df_raw["date_time"])

# After
print("Column type is: ")
print(df_raw["date_time"].dtypes)

df_raw.head(5)
```

Now the missing values.

USGS (Unites States Geological Survey) only assigns PAGER (Promt Assessment of Global Earthquakes for Response) colour values to events that directly affect population or the economy. So the missing values for "alert" can simply be filled with 'unassigned'. 

Events taking place off-shore or between territories do not get assigned a country value when recorded. These can simply be labeled as 'off-shore. Likewise, the "continent" column can be filled in as 'off-shore' or 'oceanic'. This column could be dropped entirely as we posses the latitude and longitude values of the recorded instances and they can be used for cross-referencing their geographical locaiton. Similarly, "location" can be filled with 'uknown' or be dropped.

I will choose to preserve and impute. 

```python
# Output will be truncated, view as a scrollable element
# Filling in "alert"
df_raw["alert"] = df_raw["alert"].fillna("unassigned")
print(df_raw["alert"].value_counts())
print("=" * 50)

# Filling in "country"
df_raw["country"] = df_raw["country"].fillna("off_shore")
print(df_raw["country"].value_counts())
print("=" * 50)

# Filling in "continent"
df_raw["continent"] = df_raw["continent"].fillna("off_shore")
print(df_raw["continent"].value_counts())
print("=" * 50)

# Filling in "loaction"
df_raw["location"] = df_raw["location"].fillna("off_shore")
print(df_raw["location"].value_counts())
print("=" * 50)

```

With those handled we can move on to feature engineering.

# Section 3 - Feature Engineering

We want to see how magnitude (the raw power output) behaves with its depth. Higher magnitude events, occuring at deeper depths will have different effects than ones at a shallower depth, as the energy dissipates over the time it travels. 

Multiplying magnitude by depth will give our model a combined metric to measure how much is any, depth affects the magnitude.





```python
# Making the feature
df_raw["mag_depth_interaction"] = (
    df_raw["magnitude"] * df_raw["depth"]
    ).round(2)

# View the dataframe to confirm 
df_raw.head(5)
```

Tsunami's need an earthquake to happen close the sea floor. This depth can vary, grouping those into bins makes comparing the relevance of the depth easier.

Additionally, an earthquake can't trigger a tsunami if it happens inland. An 'is_off_shore' column will act a filter for potential tsunami generation. 

```python
# "Depth" into bins
df_raw["depth_cat"] = pd.cut(
    df_raw["depth"],
    bins=[-np.inf, 70, 300, np.inf],
    labels=["shallow", "Intermediate", "Deep"],
)

# Off-shore marker
df_raw["is_off_shore"] = df_raw["country"].isin(["off_shore"])

# View to confirm the changes
df_raw.head(10)
```

Machine Learning is going to need a target label for what is a high severity event. 

We'll combine a few of the meausures that are used to classify this.
- An mmi of 7 or greater
- A sig score of 875 or greater
- An alert value of orange or red

MMI (Modified Mercalli Intensity) is used to classify the localised effect of an earthquake, rated from 1-10.

Sig score (Significance) is figure produce combining magnitude, PAGER alert level, maximum intensity, and total number of felt reports (cdi). 
We're using 875 as that's the datasets 75% percentile, creating a clear benchmark for what is high severity.

An orange alert value means 'estimated fatalities between 100 & 999 or economic losses between $100m and $1b (national scale impact) and a red alert value means 'estimated fatalties 1000+ or economic losses of over $1b (international response required).


```python
# Implimenting the rule

high_severity_alert = ["orange", "red"]

df_raw["high_severity"] = np.where(
    (df_raw["mmi"] >= 7)
    | (df_raw["sig"] >= 875)
    | df_raw["alert"].isin(high_severity_alert),
     1,
     0,
)

# View changes
df_raw.head(10)

```

Breaking down the datetime information into seperate columns will make them useful in Machine Learning for predicting patterns over seasons. Combined with location data it could also be useful for detection efforts.  

```python
# Extract datetime info into seperate columns
df_raw["year"] = df_raw["date_time"].dt.year
df_raw["month"] = df_raw["date_time"].dt.month
df_raw["hour"] = df_raw["date_time"].dt.hour

# View changes 
df_raw.head(10)
```

Day doesn't matter as earthquakes are random events. Hour is a relevant time scale for the human perception and can be used to measure the effects over its duration. Month and year help capture seasonal and annual cycle trends.
